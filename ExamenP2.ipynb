#Pregunta 3

Lee ambos CSVs en dfc (clientes) y dfo (órdenes).   
Realiza un left merge de dfc con dfo por cliente_id, manteniendo todas las filas de dfc.   
Calcula:
(a) Número total de filas del resultado del left merge.   
(b) Monto total por cliente (suma de monto) y reemplaza NaN por 0.   
Pregunta numérica:

¿Cuál es el monto total por cliente cliente_id=1?

```python
import pandas as pd
dfc = pd.read_csv('clientes.csv')
dfo = pd.read_csv('ordenes.csv')
df_merged = pd.merge(dfc, dfo, on='cliente_id', how='left')
total_filas = len(df_merged)
monto_por_cliente = df_merged.groupby('cliente_id')['monto'].sum().fillna(0)
display(df_merged)
print(f'\n(a) Número total de filas del merge: {total_filas}')
resultado_id1 = monto_por_cliente.get(1, 0)
print(f'\nMonto total para el cliente_id=1: {resultado_id1}')
```

#Pregunta 4

Carga el CSV en df, selecciona solo columnas: ts, temperatura, humedad.
Convierte ts a datetime manejando formatos mixtos (usa format='mixed' y errors='coerce').   
Reemplaza valores faltantes en temperatura con la mediana y en humedad con la media (usa fillna).   
Reporta:   
(a) ¿Cuántos NaT quedaron en ts?   
(b) Valor final de la mediana de temperatura tras imputación (redondea a 1 decimal).   
(c) Valor final de la media de humedad tras imputación (redondea a 3 decimales).   
Formato de entrega: tres números (en orden a, b, c):


```python
import pandas as pd
df_s = pd.read_csv('sensores.csv')[['ts', 'temperatura', 'humedad']]
df_s['ts'] = pd.to_datetime(df_s['ts'], format='mixed', errors='coerce')
nat_count = df_s['ts'].isna().sum()
mediana_temp = df_s['temperatura'].median()
df_s['temperatura'] = df_s['temperatura'].fillna(mediana_temp)
media_hum = df_s['humedad'].mean()
df_s['humedad'] = df_s['humedad'].fillna(media_hum)
print(f'a) NaT en ts: {nat_count}')
print(f'b) Mediana de temperatura: {round(mediana_temp, 1)}')
print(f'c) Media de humedad: {round(media_hum, 3)}')

display(df_s.head())
```
